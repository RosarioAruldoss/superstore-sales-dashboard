```python
# Importing the necessary libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Setting seaborn style
sns.set(style='whitegrid')

# Load Superstore dataset - file name: sales_data.csv
file_path = 'sales_data.csv'
df = pd.read_csv(file_path)

# to Display the first 5 rows to check
df.head()
```

```python
# to know the number of records and fields
df.shape
```

```python
# To know and understand the dataset, checking the basic information

df.info()
```

```python
df.columns
```

```python
df.describe()
```

```python
# Checking for missing values
df.isnull().sum()

# After execution, we find that in 'Postal Code' 11 values are missing
```

```python
#Identifying the missing records

missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
missing_data = pd.DataFrame({'Missing Values': missing_values, 'Percentage': missing_percentage})
missing_data = missing_data[missing_data['Missing Values'] > 0]
print(missing_data)
```

```python
# Since the percentage of records with the null values are less than 30% we can remove the records or fill the rows.
# In this case, since the data is postal code, we cannot assign any random values. Hence dropping the rows

df.dropna(subset=['Postal Code'], inplace=True)
```

```python
df.info()
#The records with the null values are removed.
```

```python
# Let's start with EDA Part 1: Sales by Region
```

```python
# Total sales by region
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

# Plotting
plt.figure(figsize=(10,6))
sns.barplot(x=region_sales.index, y=region_sales.values, palette='viridis')
plt.title('Total Sales by Region')
plt.ylabel('Sales')
plt.xlabel('Region')
plt.xticks(rotation=45)
plt.show()

# Insight: The West region has the best sales by region
```

```python
# Total sales by category:
category_profit = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

# Plotting
plt.figure(figsize=(10,6))
sns.barplot(x=category_profit.index, y=category_profit.values, palette='plasma')
plt.title('Total Profit by Product Category')
plt.ylabel('Profit')
plt.xlabel('Category')
plt.xticks(rotation=45)
plt.show()

# Insight: The category - Technology has given the most profits.
```

```python
# Total Sales by Sub-Category
plt.figure(figsize=(12,6))
subcat_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
sns.barplot(x=subcat_sales.index, y=subcat_sales.values, palette='Set3')
plt.title('Total Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
# Let's see which shipping mode is
# Count of orders by Ship Mode
ship_mode_counts = df['Ship Mode'].value_counts()

# Plotting
plt.figure(figsize=(8,8))
plt.pie(ship_mode_counts.values, labels=ship_mode_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Shipping Modes')
plt.show()
```

```python
#we encountered an error - 'Order Date' column is NOT in datetime format 
#to rectify that, change the date time format:

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

# Confirm conversion
df['Order Date'].dtypes
```

```python
# Extract month from Order Date
df['Month'] = df['Order Date'].dt.month
```

```python
# Extract month from Order Date
df['Month'] = df['Order Date'].dt.month

# Total sales per month
monthly_sales = df.groupby('Month')['Sales'].sum()

# Plotting
plt.figure(figsize=(12,6))
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values, marker='o', color='blue')
plt.title('Monthly Sales Trend')
plt.xlabel('Month (1=January, 12=December)')
plt.ylabel('Total Sales')
plt.xticks(range(1,13))
plt.grid(True)
plt.show()
```

```python
#Insight: The month of November Saw the highest sales. This trend could be due to the immenent festives and celebrations.
```
